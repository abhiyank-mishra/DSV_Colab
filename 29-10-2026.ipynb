```python
from IPython import get_ipython
from IPython.display import display
# %%
# conversation of txt into csv

with open("data.txt", "w") as f:
  f.write("Name, Department , Marks, City\n"),
  f.write("Abhiyank,CSE , 87, Auraiya\n"),
  f.write("Astik ,CSE , 83, Kanpur\n"),
  f.write("Ashutosh ,CSE , 85, Gorakhpur\n"),
  f.write("Aditya ,CSE , 82, Shampoo\n"),
# %%
with open("data.txt", "r") as f:
  data = f.read()

print(data)
# %%
import pandas as pd

df = pd.read_csv("data.txt")

df.to_csv("data.csv", index=True)

print(pd.read_csv("data.csv"))
# %%
df = pd.DataFrame({
    "Name": ["Aman","Ajay","Ravi"],
    "Mask":[90,65,87]
})
# %%
df.to_pickle("st.pkl")
display(df)
# %%
pkl = pd.read_pickle("st.pkl")
print(pkl)
# %%
import requests
from bs4 import BeautifulSoup
# %%
url = "https://www.abhiyank.in/"

page = requests.get(url)

soup = BeautifulSoup(page.text, "html.parser")

# %%
print(soup.title.text)
# %%
import sqlite3
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
print("database created successfully")
# %%
cursor.execute("""
CREATE TABLE IF NOT EXISTS DISHA (
NAME TEXT,
ROLL_NO INTEGER,
MARKS INTEGER)
""")
print("Table Creat")
# %%
#write the data into table
cursor.execute(
    "INSERT INTO DISHA VALUES(?, ?, ?)",
    ("Amit", 101, 90)
)
cursor.execute(
    "INSERT INTO DISHA VALUES('Aman', 102, 95)"
)
# %%
conn.commit()
print("records inserted successfully")
conn.close()
# %%
#retrieve data
conn = sqlite3.connect("students.db")
cursor = conn.cursor()
cursor.execute("SELECT * FROM DISHA")
rows = cursor.fetchall()
for row in rows:
    print(row)
conn.close()
# %%

```
